```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import from_unixtime, col, year, when, month, create_map, lit, date_format
from itertools import chain
import pyspark.sql.functions as F

spark = SparkSession.builder.appName('').getOrCreate()
spark.conf.set("spark.sql.legacy.parquet.nanosAsLong", "true")
spark.conf.set("spark.sql.legacy.parquet.int96RebaseModeInRead", "CORRECTED")
spark.conf.set("spark.sql.legacy.parquet.int96RebaseModeInWrite", "CORRECTED")
spark.conf.set("spark.sql.legacy.parquet.datetimeRebaseModeInRead", "CORRECTED")
spark.conf.set("spark.sql.legacy.parquet.datetimeRebaseModeInWrite", "CORRECTED")

df = spark.read.parquet("dbfs:/mnt/aviation-data/ontime.*.parquet")
df = df.withColumn("date_of_flight", date_format(from_unixtime(col("date_of_flight") / 1e9),"yyyy-MM-dd"))
df = df.withColumn("season", when(month(col("date_of_flight")).isin([12, 1, 2]), "Winter").when(month(col("date_of_flight")).isin([3, 4, 5]), "Spring").when(month(col("date_of_flight")).isin([6, 7, 8]), "Summer").otherwise("Fall"))

df = df.drop("carrier_id","diverted_airport_1", "diverted_airport_2", "diverted_airport_3", "diverted_airport_4", "diverted_airport_5", "gate_time_diverted_1", "gate_time_diverted_2", "gate_time_diverted_3", "gate_time_diverted_4", "gate_time_diverted_5", "longest_gate_time_diverted_1", "longest_gate_time_diverted_2", "longest_gate_time_diverted_3", "longest_gate_time_diverted_4", "longest_gate_time_diverted_5", "__null_dask_index__", "diverted_airport_code_2", "diverted_airport_code_3", "diverted_airport_code_4", "diverted_airport_code_5")

airlines_dict = {
    "UA": "United Airlines",
    "NK": "Spirit Airlines",
    "AA": "American Airlines",
    "B6": "JetBlue Airways",
    "DL": "Delta Air Lines",
    "F9": "Frontier Airlines",
    "HA": "Hawaiian Airlines",
    "G4": "Allegiant Air",
    "AS": "Alaska Airlines",
    "WN": "Southwest Airlines",
    "EV": "ExpressJet Airlines",
    "OO": "SkyWest Airlines",
    "VX": "Virgin America"
}

mapping_exp = create_map([lit(x) for x in chain(*airlines_dict.items())])
df = df.withColumn("airline_name", mapping_exp[col("carrier_code")])

df = df.filter(col("airline_name").isNotNull())

region_data = [(region, code) for region, codes in faa_regions.items() for code in codes]
region_df = spark.createDataFrame(region_data, ["faa_region", "arrival_airport"])

df = df.join(region_df, on="arrival_airport" , how="left")
```

```python
from pyspark.sql.functions import to_date, sum as _sum
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
import pandas as pd

# Aggregate delays by date
df_aggregated = df.groupBy(to_date("date_of_flight").alias("date")).agg(
    _sum("arrival_delay").alias("total_arrival_delay")
).orderBy("date")

# Convert to Pandas DataFrame
pandas_df = df_aggregated.toPandas()

# Set date as index and convert to time series
pandas_df["date"] = pd.to_datetime(pandas_df["date"])
pandas_df.set_index("date", inplace=True)

# Perform time series decomposition
decomposition = seasonal_decompose(pandas_df["total_arrival_delay"], model="additive", period=365)

# Plot the components
plt.figure(figsize=(12, 8))

plt.subplot(4, 1, 1)
plt.plot(decomposition.observed)
plt.title("Observed")

plt.subplot(4, 1, 2)
plt.plot(decomposition.trend)
plt.title("Trend")

plt.subplot(4, 1, 3)
plt.plot(decomposition.seasonal)
plt.title("Seasonality")

plt.subplot(4, 1, 4)
plt.plot(decomposition.resid)
plt.title("Residuals")

plt.tight_layout()
plt.show()
```

```python
from pyspark.sql.functions import dayofweek
import seaborn as sns
import matplotlib.pyplot as plt

# Extract day of week
df_with_day = df.withColumn("day_of_week", dayofweek("date_of_flight"))

# Aggregate delays by day of week
df_delay_by_day = df_with_day.groupBy("day_of_week").agg(
    _sum("arrival_delay").alias("total_arrival_delay"),
    _sum("departure_delay").alias("total_departure_delay")
).orderBy("day_of_week")

# Convert to Pandas DataFrame for visualization
pandas_delay_by_day = df_delay_by_day.toPandas()

# Map day of week numbers to names
day_map = {1: "Sunday", 2: "Monday", 3: "Tuesday", 4: "Wednesday", 5: "Thursday", 6: "Friday", 7: "Saturday"}
pandas_delay_by_day["day_of_week"] = pandas_delay_by_day["day_of_week"].map(day_map)

# Plot total delays by day of week
plt.figure(figsize=(10, 6))
sns.barplot(x="day_of_week", y="total_arrival_delay", data=pandas_delay_by_day, palette="viridis")
plt.title("Total Arrival Delay by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Total Arrival Delay (Minutes)")
plt.show()
```

```python
from pyspark.sql.functions import col, dayofweek, hour
from pyspark.ml.feature import StringIndexer, VectorAssembler
from pyspark.ml.regression import GBTRegressor
from pyspark.ml import Pipeline
from pyspark.ml.evaluation import RegressionEvaluator
import mlflow
import mlflow.spark
from pyspark.sql import functions as F

categorical_cols = ["carrier_code", "season", "faa_region"]
indexers = [
        StringIndexer(inputCol=column, outputCol=f"{column}_indexed", handleInvalid="skip")
        for column in categorical_cols
]

feature_cols = [
    "day_of_week", "departure_delay", "arrival_delay", "elapsed_time_difference", "carrier_delay", 
    "weather_delay", "nas_delay", "security_delay", "late_aircraft_delay", "ground_time_from_gate",
    "longest_gate_time"
] + [f"{col}_indexed" for col in categorical_cols]

assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")    

gbt = GBTRegressor(
    labelCol="arrival_delay",
    featuresCol="features",
    maxIter=50,
    maxDepth=5,
    maxBins=392,
    seed=42
)

# Create pipeline
pipeline = Pipeline(stages=indexers + [assembler, gbt])

# Split data
train_data, test_data = df.randomSplit([0.7, 0.3], seed=42)

with mlflow.start_run():
    # Log parameters
    mlflow.log_param("maxIter", 50)
    mlflow.log_param("maxDepth", 5)
    mlflow.log_param("seed", 42)

    # Train the model
    model = pipeline.fit(train_data)

    # Make predictions
    predictions = model.transform(test_data)

    # Evaluate the model
    evaluator = RegressionEvaluator(
        labelCol="arrival_delay",
        predictionCol="prediction",
        metricName="rmse"
    )
    rmse = evaluator.evaluate(predictions)
    print(f"Root Mean Squared Error (RMSE): {rmse}")

    # Log metrics
    mlflow.log_metric("rmse", rmse)

    # Log the model
    mlflow.spark.log_model(model, "gbt_model")

    # Save the model (optional)
    model.save("/dbfs/flight_pred/gbt_model")
```
